# Quick Start: Load Pre-computed Embeddings

This notebook loads your pre-computed vector store without re-running embeddings.


```python
# Import the helper functions
from load_embeddings import load_vector_store, retrieve_texts

# Load your pre-computed vector store
vs = load_vector_store()
```

```python
# Test retrieval
if vs is not None:
    query = "acts related to revenue in Bombay"
    results = retrieve_texts(vs, query, k=3)
    
    for i, r in enumerate(results):
        print(f"\n--- Result {i+1} (Score: {r['score']:.3f}) ---")
        print(f"Title: {r['metadata'].get('Short Title', 'N/A')}")
        print(f"Text preview: {r['text'][:200]}...")
```

```python
# Now you can search for any legal question
def search_legal_question(question, k=5):
    """Search for legal information."""
    if vs is None:
        print("❌ Vector store not loaded!")
        return []
    
    results = retrieve_texts(vs, question, k)
    print(f"🔍 Found {len(results)} relevant results for: '{question}'")
    
    for i, r in enumerate(results):
        print(f"\n--- Result {i+1} (Score: {r['score']:.3f}) ---")
        print(f"Title: {r['metadata'].get('Short Title', 'N/A')}")
        print(f"Date: {r['metadata'].get('Enactment Date', 'N/A')}")
        print(f"Act Number: {r['metadata'].get('Act Number', 'N/A')}")
        print(f"Entity: {r['metadata'].get('Entity', 'N/A')}")
        print(f"Text preview: {r['text'][:300]}...")
    
    return results

# Example usage
search_legal_question("What are the provisions for taxation laws?")
```
